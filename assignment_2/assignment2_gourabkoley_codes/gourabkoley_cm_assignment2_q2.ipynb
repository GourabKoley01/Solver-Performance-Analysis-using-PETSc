Given $P(x)$ = $ax^2$ + $bx $ + $c$ $\ $ for the function $f(x)$ = $2^x$  at the points $x_{i} $ = 0, 1, 2, 3, 4, so seting up the $P(x_{i})$ and $f(x_{i})$

|x | $f(x) = 2^x$ |
|----------|----------|
| 0 | 1 | 
| 1 | 2 | 
| 2 | 4 |
| 3 | 8 |
| 4 | 16 |

```python
x = np.array([0, 1, 2, 3, 4])
y = 2 ** x
n = len(x)
```

The goal is to minimize $E(a, b, c) $ = $ \sum_{i = 0}^4 $  $[2^{x_{i}} - (ax_{i}^2 + bx_{i} + c)  ]^2$ 

Taking the partial derivative with respect to $a, b, c$ set eacch to zero to obtain three linear equation, which produces 

$$
\begin{align}
\sum (ax_{i}^2 + bx_{i} + c ) &= \sum 2^x_{i} \\
\sum x_{i}(ax_{i}^2 + bx_{i} + c ) &= \sum x_{i} 2^{x_{i}} \\
\sum x_{i}^2 (ax_{i}^2 + bx_{i} + c ) &= \sum x_{i}^2 2^{x_{i}}
\end{align}
$$

Which can be written as 

$$
\begin{align}
a \sum x_{i}^4 + b \sum x_{i}^3 + c \sum x_{i}^2 &= \sum x_{i}^2 2^x_{i} \\
a \sum x_{i}^3 + b \sum x_{i}^2 + c \sum x_{i}^1 &= \sum x_{i} 2^x_{i} \\
a \sum x_{i}^2 + b \sum x_{i} + cn &= \sum  2^x_{i} \\
\end{align}
$$

When plugged in into the equation solves for $a, b$  and $c$

```python
A = np.array([
    [np.sum(x**4), np.sum(x**3), np.sum(x**2)],
    [np.sum(x**3), np.sum(x**2), np.sum(x)], 
    [np.sum(x**2), np.sum(x), n ]
])
B = np.array([
    np.sum((x**2) * y), 
    np.sum(x * y), 
    np.sum(y)
])

coefficient = np.linalg.solve(A,B)
a, b, c = coefficient

print("A : ", a)
print("B : ", b)
print("C : ", c)
print(f"Least square quadretic fit : P(x) = {a} x^2 + {b} x + {c}")
```
